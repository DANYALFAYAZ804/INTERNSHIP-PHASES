```python
secret_number = 4
guess = 0

guess = int(input("Please enter your guessed number: "))

while guess != secret_number:
	if guess < secret_number:
		print("Too low")
	elif guess > secret_number:
		print("Too high")
	else:
		print("Congratulations! You guessed it!")
```

```python
# 1. Rectangle Area Calculator
def rectangle_area(length, width):
    area = length * width
    return area

# 2. Minutes to Hours and Minutes
def convert_minutes(total_minutes):
    hours = total_minutes // 60
    minutes = total_minutes % 60
    return str(hours) + "h " + str(minutes) + "m"

# 3. Split the Restaurant Bill
def split_bill(bill, tip_percent, people):
    tip_amount = bill * (tip_percent / 100)
    total = bill + tip_amount
    return total / people

# 4. Even Number Check
def is_even(number):
    return number % 2 == 0

# 5. Number Sign Checker
def check_sign(num):
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"

# 6. Movie Ticket Pricer
def ticket_price(age):
    if age < 5:
        return 0
    elif age <= 17:
        return 10
    elif age <= 64:
        return 15
    else:
        return 12

# 7. Maximum of Three Numbers
def max_of_three(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

# 8. Leap Year Validator
def is_leap_year(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    else:
        return False

# 9. Sum from 1 to N
def sum_to_n(n):
    total = 0
    for i in range(1, n + 1):
        total = total + i
    return total

# 10. Factorial Calculator
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result = result * i
    return result

# 11. Count Multiples in a Range
def count_multiples(start, end, divisor):
    count = 0
    for i in range(start, end + 1):
        if i % divisor == 0:
            count = count + 1
    return count

# 12. Digit Counter with a While Loop
def count_digits(n):
    if n == 0:
        return 1
    count = 0
    while n > 0:
        n = n // 10
        count = count + 1
    return count
```
